```python
import pandas as pd
import numpy as np
from nltk.stem.porter import PorterStemmer
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import pickle
from fuzzywuzzy import fuzz
```

```python
speakers = pd.read_csv("Speakers.csv")
speakers.drop(['main_category', 'sub_category'], axis=1, inplace=True)
```

```python
speakers['#keywords'] = speakers['name'].str.lower()
speakers['#keywords'] = speakers["#keywords"].str.replace(r'[^\w\s]', ' ', regex=True)
```

```python
stemmer = PorterStemmer()
def stemming(text):
    words = []
    for word in text.split(' '):
        words.append(stemmer.stem(word))
    return ' '.join(words)

speakers['#keywords'] = speakers['#keywords'].apply(stemming)
```

```python
cv = CountVectorizer(max_features=5000, stop_words='english', dtype=np.uint8)
cv.fit(speakers['#keywords'])
vector = cv.transform(speakers['#keywords']).toarray()
```

```python
similarity = cosine_similarity(vector)
del vector
```

```python
def find_closest_product(search_text, df):

    max_ratio = 0
    closest_match = None
    

    search_text = search_text.lower()
    
    for product_name in df['name']:

        ratio = fuzz.partial_ratio(search_text.lower(), product_name.lower())
        if ratio > max_ratio:
            max_ratio = ratio
            closest_match = product_name
    
    return closest_match if max_ratio > 50 else None
```

```python
def search_and_recommend(search_text, df=speakers, similarity_matrix=similarity, top_n=10):
    
    matched_product = find_closest_product(search_text, df)
    
    if matched_product is None:
        return None, []
    

    product_index = df[df['name'] == matched_product].index[0]
    similarity_list = list(enumerate(similarity_matrix[product_index]))
    top_similar_products = sorted(similarity_list, key=lambda x: x[1], reverse=True)[1:top_n+1]
    
    recommendations = []
    for idx, similarity_score in top_similar_products:
        recommendations.append({
            'product': df.loc[idx]['name'],
            'similarity_score': similarity_score
        })
    
    return matched_product, recommendations
```

```python

pickle.dump(similarity, open('similarity.pkl', 'wb'))
pickle.dump(speakers, open('data.pkl', 'wb'))
```

```python
def search_products(search_text):

    matched_product, recommendations = search_and_recommend(search_text)
    
    if matched_product is None:
        print(f"No products found matching '{search_text}'")
        return
    
    print(f"\nBest match found: {matched_product}")
    print("\nRecommended similar products:")
    for i, rec in enumerate(recommendations, 1):
        print(f"{i}. {rec['product']} (Similarity: {rec['similarity_score']:.2f})")
```
